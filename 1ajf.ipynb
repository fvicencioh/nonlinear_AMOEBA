```python
from coupled_solver.solver import solver

ep_in = 4.
ep_ex = 78.3
k = 0.125

ajf = solver('1ajf/1ajf', ep_in, ep_ex, k)
ajf.solve_vacuum(use_record_info=True)
ajf.solve_system(grid_scale=4., grid_scale_ratio=4., non_linear=True)
```

```python
ajf.export_potential('surface',filename='1ajf_surface_nonlinear.vtu')
ajf.export_potential('solute', filename='1ajf_solute_nonlinear.vtu')
ajf.export_potential('volume', filename='1ajf_volume_nonlinear.vtu')
```

```python
from coupled_solver.solver import solver

ep_in = 4.
ep_ex = 78.3
k = 0.125

ajf = solver('1ajf/1ajf', ep_in, ep_ex, k)
ajf.solve_vacuum()
ajf.solve_system(grid_scale=4., grid_scale_ratio=4., non_linear=False)
```

```python
ajf.export_potential('surface',filename='1ajf_surface_linear.vtu')
ajf.export_potential('solute', filename='1ajf_solute_linear.vtu')
ajf.export_potential('volume', filename='1ajf_volume_linear.vtu')
```
